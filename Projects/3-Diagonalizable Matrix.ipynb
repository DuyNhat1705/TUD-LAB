# Thông tin giải thuật

- Giải thuật chéo hóa ma trận có cội nguồn từ thế kỷ 18 và 19, gắn liền với các bài toán vật lý như cơ học thiên thể và dao động của hệ nhiều vật. Các nhà toán học như Leonhard Euler và Joseph-Louis Lagrange đã đặt những nền móng đầu tiên khi nghiên cứu sự quay của vật thể cứng và chuyển động của các hành tinh. Sau đó, Augustin-Louis Cauchy là người chính thức đưa ra khái niệm trị riêng, vectơ riêng và chứng minh rằng các ma trận đối xứng thực luôn có thể chéo hóa được. Đến nửa sau thế kỷ 19, Arthur Cayley hoàn thiện lý thuyết đại số ma trận, biến phép chéo hóa từ một công cụ giải phương trình vi phân trong vật lý trở thành một giải thuật nền tảng của đại số tuyến tính hiện đại.

- Chéo hóa ma trận là quá trình tìm một ma trận khả nghịch $P$ và một ma trận đường chéo $D$ sao cho ma trận vuông $A$ cấp $n$ được biểu diễn dưới dạng $A = P  D  P^{-1}$. Trong đó, $D$ chứa các giá trị riêng trên đường chéo chính và $P$ gồm các cột là các vectơ riêng tương ứng. Một ma trận $A$ chéo hóa được khi và chỉ khi nó có đủ $n$ vectơ riêng độc lập tuyến tính.

- Thuật toán chéo hóa gồm 3 bước cơ bản:
  - **Tìm các trị riêng:** Giải phương trình đặc trưng $\det(A - \lambda I) = 0$ để tìm các giá trị $\lambda$.

  - **Tìm vectơ riêng & Kiểm khả năng chéo hóa:** Giải hệ phương trình thuần nhất $(A - \lambda I)v = 0$ cho từng $\lambda$ để tìm cơ sở không gian riêng. Sau đó, xác định xem tổng số vectơ riêng độc lập tuyến tính thu được có bằng cấp $n$ của ma trận hay không.

  - **Xây dựng $P$ và $D$:** Ghép các vectơ riêng thành các cột của ma trận $P$, tìm nghịch đảo của $P$ - $P^{-1}$ và tìm ma trận làm chéo D qua công thức $D = P^{-1}AP$.

```python
import sympy as sp

def printMatrix(M):
    for row in M:
        print(row)

def relativeCompare(num1, num2, eps = 1e-9):
    return abs(num1 - num2) <= eps

def unitMatrix(n):
    return [[1 if i == j else 0 for j in range(n)] for i in range(n)]


def connectMatrixWithCoefficientVector(M,v):
    n = len(M)
    if n != len(v):
        raise ValueError("Ma trận và vector phải có cùng số hàng.")
    for i in range(n):
        M[i].append(v[i])
    return M

def connectMatrixWithMatrix(M1, M2):
    M = M1
    n = len(M1)
    if n != len(M2):
        raise ValueError("Ma trận phải có cùng số hàng để kết nối.")
    for i in range(n):
        M[i].extend(M2[i])
    return M
```

```python
def swap_row(M, i, j):
    M[i],M[j] = M[j],M[i]

def multiply(M, i, alpha):
    n = len(M[i])
    for x in range (n):
        M[i][x] = M[i][x]*alpha

def eliminate_row(M, i, j, k):
    n = len(M[i])
    for x in range (n):
        val = M[j][x] - k*M[i][x]
        M[j][x] = 0 if relativeCompare(val,0) else val
```

```python
def multiplyMatrixWithNumber(M,a):
    n = len(M)
    m = len(M[0])
    result = []
    for i in range(n):
        row_result = []
        for j in range(m):
            row_result.append(M[i][j] * a)
        result.append(row_result)
    return result

def plusMatrix(M1,M2):
    n = len(M1)
    m = len(M1[0])
    if n != len(M2) or m != len(M2[0]):
        raise ValueError("Ma trận phải có cùng kích thước để cộng.")
    result = []
    for i in range(n):
        row_result = []
        for j in range (m):
            row_result.append(M1[i][j]+M2[i][j])
        result.append(row_result)
    return result

def multiplyMatrixWithMatrix(M1,M2):
    n = len(M1)
    m = len(M2[0])
    p = len(M2)
    if len(M1[0]) != p:
        raise ValueError("Số cột trong M1 phải bằng số hàng trong M2.")
    result = []
    for i in range(n):
        row_result = []
        for j in range(m):
            sum_product = 0
            for k in range(p):
                sum_product += M1[i][k] * M2[k][j]
            if relativeCompare(sum_product, 0):
                sum_product = 0
            row_result.append(sum_product)
        result.append(row_result)
    return result
```

```python
def removeColumnRow(M, col, row):
    result = []
    n = len (M)
    m = len (M[0])
    for i in range(n):
        if i != row:
            row_list = []
            for j in range(m):
                if j != col:
                    row_list.append(M[i][j])
            result.append(row_list)
    return result

def det(M):
    n = len(M)
    m = len(M[0])
    if(n != m):
        raise ValueError("Matrix must be square")
    if(n == 1):
        return M[0][0]
    result = 0
    for j in range(m):
        result += M[0][j] * (-1)**j * det(removeColumnRow(M, j, 0))
    return result
```

```python
def solveLinearEquation(expr, symbols):
    poly = sp.Poly(expr, symbols)
    coeffs = poly.all_coeffs()
    
    a, b = coeffs[0], coeffs[1]
    if a == 0:
        return []
    
    return [sp.simplify(-b / a)]

def solveQuadraticEquation(expr, symbols):
    poly = sp.Poly(expr, symbols)
    coeffs = poly.all_coeffs()
    
    while len(coeffs) < 3:
        coeffs.insert(0, 0)
        
    a, b, c = coeffs[0], coeffs[1], coeffs[2]
    if a == 0:
        return solveLinearEquation(expr, symbols)

    delta = b**2 - 4 * a * c
    delta_simplified = sp.simplify(delta)

    # 1. Delta < 0 -> Vô nghiệm thực
    if delta_simplified < 0:
        return []
    
    # 2. Delta = 0 -> 1 nghiệm kép
    elif delta_simplified == 0:
        return [sp.simplify(-b / (2 * a))]
    
    # 3. Delta > 0 -> 2 nghiệm thực phân biệt
    else:
        r1 = sp.simplify((-b + sp.sqrt(delta)) / (2 * a))
        r2 = sp.simplify((-b - sp.sqrt(delta)) / (2 * a))
        return [r1, r2]

def solveCubicEquation(expr, symbols):

    poly = sp.Poly(expr, symbols)
    coefficients = poly.all_coeffs()

    while len(coefficients) < 4:
        coefficients.insert(0, 0)

    a, b, c, d = coefficients[0], coefficients[1], coefficients[2], coefficients[3]

    if a == 0:
        return solveQuadraticEquation(expr, symbols)

    try:
        factored = sp.factor(expr)
        if factored != expr:  # Nếu phân tích nhân tử thành công
            factors = (
                factored.args if isinstance(factored, sp.Mul) else [factored]
            )
            rational_roots = []
            quadratic_part = None

            for factor in factors:
                # Xử lý trường hợp có mũ, ví dụ (x - 2)**2
                if factor.is_Pow:
                    base, exp = factor.as_base_exp()
                    if base.has(symbols):
                        deg = sp.degree(base, symbols)
                        if deg == 1:
                            rational_roots.extend(
                                sp.solve(base, symbols) * int(exp)
                            )
                elif factor.has(symbols):
                    deg = sp.degree(factor, symbols)
                    if deg == 1:
                        rational_roots.extend(sp.solve(factor, symbols))
                    elif deg == 2:
                        quadratic_part = factor

            # Nếu tìm thấy nhân tử bậc 1 -> Giải nốt bậc 2 bằng hàm sẵn có
            if len(rational_roots) >= 1 and quadratic_part is not None:
                q_roots = solveQuadraticEquation(quadratic_part, symbols)
                all_found = rational_roots + q_roots
                # Làm sạch và lọc nghiệm trùng
                clean_roots = []
                for r in all_found:
                    val = r
                    if not any(
                        relativeCompare(val, existing)
                        for existing in clean_roots
                    ):
                        clean_roots.append(val)
                return clean_roots

            elif len(rational_roots) >= 3:
                clean_roots = []
                for r in rational_roots:
                    val = r
                    if not any(
                        relativeCompare(val, existing)
                        for existing in clean_roots
                    ):
                        clean_roots.append(val)
                return clean_roots
    except Exception:
        pass  # Nếu factor gặp lỗi, tự động rơi xuống Cardano bên dưới

    b, c, d = b / a, c / a, d / a
    a = 1

    p = (3 * a * c - b**2) / (3 * a**2)
    q = (2 * b**3 - 9 * a * b * c + 27 * a**2 * d) / (27 * a**3)

    delta_c = sp.simplify((q / 2) ** 2 + (p / 3) ** 3)
    shift = b / (3 * a)

    # --- Trường hợp Delta_c = 0 (Nghiệm kép/bội) ---
    if relativeCompare(delta_c, 0):
        # Dùng Rational(1, 3) thay vì cbrt để tránh lỗi nhánh phức
        u = (-q / 2) ** sp.Rational(1, 3)
        if relativeCompare(u, 0):
            # 1 Nghiệm bội 3
            return [-shift]
        else:
            # 1 Nghiệm đơn, 1 Nghiệm kép
            t1 = 2 * u
            t2 = -u
            r1 = t1 - shift
            r2 = t2 - shift
            return [r1, r2] if not relativeCompare(r1, r2) else [r1]

    # --- Trường hợp Delta_c != 0 ---
    val_u = sp.simplify(-q / 2 + sp.sqrt(delta_c))
    val_v = sp.simplify(-q / 2 - sp.sqrt(delta_c))

    u = (val_u) ** sp.Rational(1, 3)
    v = (val_v) ** sp.Rational(1, 3)

    w = sp.exp(2 * sp.I * sp.pi / 3)
    w_sq = w**2

    all_roots = [
        sp.simplify(u + v - shift),
        sp.simplify(u * w + v * w_sq - shift),
        sp.simplify(u * w_sq + v * w - shift),
    ]

    # Lọc nghiệm thực phân biệt
    real_roots = []
    for root in all_roots:
        root_expanded = sp.expand(root, complex=True)

        if sp.im(root_expanded) == 0 or not root.has(sp.I):
            val = sp.re(root_expanded)

            if not any(
                relativeCompare(val, existing) for existing in real_roots
            ):
                real_roots.append(val)

    return real_roots

def solvePolynomialEquation(expr, symbols):
    degree = sp.degree(expr, symbols)
    if degree == 1:
        return solveLinearEquation(expr, symbols)
    elif degree == 2:
        return solveQuadraticEquation(expr, symbols)
    elif degree == 3:
        return solveCubicEquation(expr, symbols)
    else:
        return sp.solve(expr, symbols)
```

```python
def GaussJordan_elimination(M):
    num_rows = len(M)
    num_cols = len(M[0])
    
    pivot_row = 0
    pivot_col = 0
    
    while pivot_row < num_rows and pivot_col < num_cols:
        # 1. Tìm cột trái nhất không chứa toàn số 0
        nonzero_row = pivot_row
        while nonzero_row < num_rows and relativeCompare(M[nonzero_row][pivot_col],0):
            nonzero_row += 1

        if nonzero_row == num_rows:
            pivot_col += 1
            continue
        # 2. Hoán vị dòng chứa phần tử khác 0 lên dòng pivot hiện tại
        if nonzero_row != pivot_row:
            swap_row(M, pivot_row, nonzero_row)
        
        # 3. Nếu giá trị được kiểm bằng 0 thì gọi multiply để chuẩn hoá về 1
        pivot_val = M[pivot_row][pivot_col]
        if not relativeCompare(pivot_val,1):
            multiply(M, pivot_row, 1/pivot_val)
        
        # 4. Duyệt tất cả các dòng khác dòng chốt pivot_row trong cột chốt pivot_col để chuẩn hoá về 0
        for x in range (0,num_rows):
            if x == pivot_row:
                continue
            if not relativeCompare(M[x][pivot_col], 0):
                eliminate_row(M, pivot_row, x, M[x][pivot_col])
        
        pivot_row += 1
        pivot_col += 1

def back_substitution(M):
    num_rows = len(M)
    num_cols = len(M[0])
    num_vars = num_cols - 1

    result = [None]*num_vars
    
    pivot_positions = {}
    i = num_rows - 1
    # 1. Tìm phần tử cơ sở cho các dòng trong ma trận M
    while i >= 0:
        # 1.1 Tìm phần tử cơ sở trong dòng i
        pivot_col = -1
        for j in range(num_vars):
            if not relativeCompare(M[i][j],0):
                pivot_col = j
                break
        
        # 1.2 Nếu không tìm thấy phần tử cơ sở và phần tử cuối cùng của dòng i khác 0, hệ phương trình vô nghiệm
        if pivot_col == -1 and not relativeCompare(M[i][num_cols - 1],0):
            print("Hệ phương trình vô nghiệm")
            return result
        
        # 1.3 Lưu lại nếu tìm thấy phần tử cơ sở của dòng i
        if pivot_col != -1:
            pivot_positions[i] = pivot_col
        i-=1
        
    # 2. Gán ẩn số thực tự do cho biến tương ứng với phần tử không nằm trong pivot_cols
    symbols = ['t', 's', 'u', 'v', 'x', 'y', 'z']
    sym = 0
    pivot_cols = set(pivot_positions.values())
    for j in range (num_vars):
        if j not in pivot_cols:
            result[j] = sp.symbols(symbols[sym])
            sym+=1
    
    i = num_rows - 1
    # 3. Duyệt ngược từ cuối lên đầu để giải hệ phương trình
    while i >= 0:
        # 3.1 Nếu dòng i không có phần tử cơ sở, bỏ qua
        if i not in pivot_positions:
            i-=1
            continue
        # 3.2 Giải phương trình dòng i để tìm ẩn số tương ứng với phần tử cơ sở
        j = pivot_positions[i]

        result[j] = M[i][num_cols - 1]

        for k in range(j + 1, num_vars):
            result[j] -= M[i][k] * result[k]
        i-=1
        
    return result

def null(M):
    GaussJordan_elimination(M)
    vectorZero = [0]*len(M)
    M = connectMatrixWithCoefficientVector(M, vectorZero)
    return back_substitution(M)

def getBasisVectors(null_space_expr):
    
    # Tìm tất cả các biến tự do (symbols) có trong null_space
    free_vars = set()
    for expr in null_space_expr:
        if isinstance(expr, sp.Basic):
            free_vars.update(expr.free_symbols)
    
    if not free_vars:
        return [null_space_expr]
    
    basis_vectors = []
    for target_var in free_vars:
        # Tải bảng giá trị: biến hiện tại = 1, các biến khác = 0
        subs_dict = {var: (1 if var == target_var else 0) for var in free_vars}

        vec = []
        for expr in null_space_expr:
            # KIỂM TRA: Nếu expr là SymPy -> dùng .subs(), nếu là số int/float -> giữ nguyên
            if isinstance(expr, sp.Basic):
                val_sub = expr.subs(subs_dict)
            else:
                val_sub = expr
            vec.append(val_sub)
        basis_vectors.append(vec)
    
    return basis_vectors
```

```python
def getPMatrix(combinedBasis):
    return [[combinedBasis[j][i] for j in range(len(combinedBasis))] for i in range(len(combinedBasis[0]))]

def inverseMatrix(A):
    M = connectMatrixWithMatrix(A, unitMatrix(len(A)))
    num_rows = len(M)
    
    # Ma trận A phải là ma trận vuông, nên M phải có số cột = 2 * số dòng
    n = num_rows 
    
    pivot_row = 0
    pivot_col = 0
    
    # CHỈ DUYỆT CÁC CỘT CỦA MA TRẬN A (từ cột 0 đến n - 1)
    while pivot_row < n and pivot_col < n:
        
        # 1. Tìm cột trái nhất không chứa toàn số 0 trong nửa trái (A)
        nonzero_row = pivot_row
        while nonzero_row < num_rows and relativeCompare(M[nonzero_row][pivot_col], 0):
            nonzero_row += 1

        # Nếu cả cột ở nửa trái đều bằng 0 => Ma trận A không nghịch đảo được
        if nonzero_row == num_rows:
            print(f"Lỗi: Ma trận không có ma trận nghịch đảo!")
            return None
            
        # 2. Hoán vị dòng chứa phần tử khác 0 lên dòng pivot hiện tại
        if nonzero_row != pivot_row:
            swap_row(M, pivot_row, nonzero_row)
        
        # 3. Chuẩn hoá phần tử chốt (pivot) về 1
        pivot_val = M[pivot_row][pivot_col]
        if not relativeCompare(pivot_val, 1):
            multiply(M, pivot_row, 1 / pivot_val)
        
        # 4. Khử tất cả các dòng khác trong cột pivot_col về 0 (trên cả nửa A và I)
        for x in range(0, num_rows):
            if x == pivot_row:
                continue
            if not relativeCompare(M[x][pivot_col], 0):
                eliminate_row(M, pivot_row, x, M[x][pivot_col])
        
        pivot_row += 1
        pivot_col += 1

    # 5. Tách ma trận nghịch đảo từ nửa phải của M
    A_inv = []
    for r in range(n):
        # Lấy từ cột n đến hết (cột 2n - 1)
        A_inv.append(M[r][n:])
        
    return A_inv

def getDiagonalMatrix(P, A, inverseP):
    return multiplyMatrixWithMatrix(multiplyMatrixWithMatrix(inverseP, A), P)
```

```python
def diagonalizeMatrix(M):
    # 4.1. Tìm các trị riêng (eigenvalues) của ma trận M
    I = unitMatrix(len(M))
    x = sp.symbols('x')
    M_var = plusMatrix(multiplyMatrixWithNumber(I, -x), M)
    expr = sp.expand(det(M_var))
    roots = solvePolynomialEquation(expr, x)

    # 4.2. Tìm các vector riêng (eigenvectors) tương ứng với từng trị riêng
    combinedBasis = []
    for i in range(len(roots)):
        A = plusMatrix(M, multiplyMatrixWithNumber(I, -roots[i]))
        nullSpace = null(A)
        basisVectors = getBasisVectors(nullSpace)
        combinedBasis.extend(basisVectors)
    if len(combinedBasis) < len(M):
        print("Ma trận không chéo hóa được")
        return None, None, None

    # 4.3. Tạo ma trận P từ các vector riêng, tính ma trận nghịch đảo của P, và tính ma trận chéo D
    P = getPMatrix(combinedBasis)
    P_tmp = getPMatrix(combinedBasis)
    inverseP = inverseMatrix(P_tmp)
    D = getDiagonalMatrix(P, M, inverseP)
    
    print("Ma trận P:")
    printMatrix(P)
    print("\nMa trận nghịch đảo P^-1:")
    printMatrix(inverseP)
    print("\nMa trận đường chéo D:")
    printMatrix(D)
    return P, inverseP, D
```

# Ý tưởng thực hiện và mô tả các hàm
## 1. Tiện ích
- `sympy`: Thư viện này hỗ trợ trong việc biểu diễn đa thức đặc trưng có được từ định thức của ma trận $A - \lambda I$.

- `printMatrix(M)`: Hàm này dùng để in ma trận $M$ theo format đẹp và cân đối hơn.

- `relativeComparison(a,b,eps = 1e-9)`: Hàm này dùng để so sánh bằng giữa hai số $a$, $b$ bằng cách so sánh hiệu của chúng xem có bé hơn hoặc bằng sai số cực nhỏ $eps$ cho trước hay không. Thay vì dùng so sánh `==` truyền thống, có thể dẫn đến kết quả sai do vấn đề làm tròn số hoặc biểu diễn dấu chấm động, việc so sánh sử dùng hàm này sẽ đảm bảo kết quả đúng khi thao tác với số thực.

- `unitMatrix(n)`: Hàm này trả về ma trận đơn vị $I$ tương ứng với kích $n$ truyền vào.

- `connectMatrixWithCoefficientVector(M,v)`: Hàm này dùng để ghép ma trận $M$ và vector $v$ đầu vào để tạo ra ma trận mở rộng dạng $(M|v)$.

- `connectMatrixWithMatrix(M1, M2)`: Hàm này dùng để ghép hai ma trận $M1$ và $M2$ đầu vào để tạo ra ma trận mở rộng dạng $(M1|M2)$.

## 2. Các phép biến đổi sơ cấp trên dòng
Các phép biến đổi sơ cấp trên dòng được ở đồ án này được dùng bởi **giải thuật Gauss-Jordan**, để tìm không gian riêng và ma trận nghịch đảo.

- `swap_row(M, i, j)`: Hàm này nhận 3 tham số, ma trận là $M$, chỉ số hàng i là $i$ và chỉ số hàng j là $j$, và thực hiện hoán vị dòng i và dòng j trong ma trận M.

- `multiply(M, i, alpha)`:  Hàm này nhận 3 tham số, ma trận là $M$, chỉ số hàng i là $i$ và một số thực $alpha$, và thực hiện nhân số thực alpha vào từng phần tử trong dòng i của ma trận M.

- `eliminate_row(M, i, j, k)`: Hàm này 4 tham số, ma trận là $M$, chỉ số hàng i là $i$, chỉ số hàng j là $j$ và hệ số $k$, và thực hiện tính toán giá trị mới cho $dòng_j$ theo công thức: $dòng_j = dòng_j - k\times dòng_i$.

## 3. Các phép toán trên ma trận
-  `multiplyMatrixWithNumber(M,a)`: Hàm này nhận vào ma trận $M$, số thực $a$ và thực hiện nhân từ phần tử của ma trận $M$ với số thực $a$ đó rồi trả về ma trận kết quả $result$ $(result_{ij} = M_{ij} \times a)$.

- `plusMatrix(M1,M2)`: Hàm này nhận vào ma trận $M1$, ma trận $M2$ và thực hiện phép công 2 ma trận - cộng các phần tử ở vị trí tương ứng trong 2 ma trận này rồi trả về ma trận kết quả $result$ $(result_{ij} = M1_{ij} + M2_{ij})$. Trước khi thực hiện phép cộng ma trận, phải kiểm tra $M1$ và $M2$ để đảm bảo kích thước của 2 ma trận bằng nhau.

- `multiplyMatrixWithMatrix(M1,M2)`: Hàm này nhận ma trận $M1$, ma trận $M2$ và thực hiện phép nhân ma trận $M1 \times M2 $và trả về ma trận kết quả $result$, với mỗi phần tử được tính theo công thức $result_{ij} = \sum_{k=1}^{p} M1_{ik} \cdot M2_{kj}$. Trước khi thực hiện phép nhân ma trận, phải kiểm tra $M1$ và $M2$ để đảm bảo số dòng của ma trận $M1$ bằng số dòng của ma trận $M2$ $(len(M1[0]) == len(M2))$.

## 4. Giải thuật chéo hóa
Giải thuật chéo hóa bao gồm 3 giai đoạn **Tìm các trị riêng của ma trận**, **Tìm vector riêng & Kiểm tra điều kiện chéo hóa** và **Xây dựng ma trận $P$, $P^{-1}$ và ma trận chéo $D$**.
### 4.1. Tìm các trị riêng của ma trận
- `removeColumnRow(M, col, row)`: Hàm này nhận vào 3 tham số là ma trận $M$, chỉ số cột $col$, chỉ số dòng $row$ và thực hiện việc xóa cột $col$ và dòng $row$ trong ma trận M. Hàm này sẽ được gọi trong hàm tính định thức `det(M)` để phục vụ công thức **Laplace**.

- `det(M)`: Hàm này nhận vào ma trận $M$ và tính định thức của ma trận theo công thức **Laplace** $\det(M) = \sum_{j=1}^{n} (-1)^{j} \cdot M_{0j} \cdot \det(M(0 | j)$ được khai triển theo dòng đầu tiên có chỉ só $i = 0$. Trước khi tính, phải kiểm tra $M$ để đảm bảo $M$ là ma trận vuông $(len(M) == len(M[0]))$.

- `solvePolynomialEquation(expr, symbols)`: Hàm đóng vai trò điều hướng chính để giải đa thức đặc trưng $expr$ theo biến số $symbols$. Dựa vào bậc của đa thức (`sp.degree`), hàm sẽ tự động gọi `solveLinearEquation` cho bậc 1, `solveQuadraticEquation` cho bậc 2, `solveCubicEquation` cho bậc 3 và sử dụng `sp.solve` của `sympy` làm phương án dự phòng cho các đa thức bậc cao hơn 3.

  - `solveLinearEquation(expr, symbols)`: Hàm giải phương trình bậc nhất $ax + b = 0$. Sau khi trích xuất các hệ số, nếu hệ số $a = 0$ hàm sẽ trả về danh sách rỗng (vô nghiệm). Ngược lại, nghiệm duy nhất được tính đơn giản theo công thức $x = -\frac{b}{a}$.

  - `solveQuadraticEquation(expr, symbols)`: Hàm giải phương trình bậc hai $ax^2 + bx + c = 0$. Hàm tự động bổ sung hệ số $0$ nếu đa thức bị khuyết bậc và sẽ chuyển về giải bậc nhất nếu $a = 0$. Dựa vào biệt thức $\Delta = b^2 - 4ac$, hàm trả về danh sách rỗng nếu $\Delta < 0$, trả về 1 nghiệm kép $x = -\frac{b}{2a}$ nếu $\Delta = 0$, hoặc 2 nghiệm thực phân biệt $x_{1,2} = \frac{-b \pm \sqrt{\Delta}}{2a}$ nếu $\Delta > 0$.

  - `solveCubicEquation(expr, symbols)`: Hàm giải phương trình bậc ba $ax^3 + bx^2 + cx + d = 0$ bằng phương pháp kết hợp. Đầu tiên, hàm ưu tiên dùng `sp.factor` để phân tích nhân tử tìm các nghiệm đại số đẹp nhằm tránh sai số. Nếu không phân tích được, hàm chuyển sang công thức **Cardano**: xử lý riêng trường hợp $\Delta_c = 0$ để tìm nghiệm bội/đơn chính xác, và dùng công thức tổng quát với số phức **Euler** $e^{i\frac{2\pi}{3}}$ khi $\Delta_c \neq 0$. Cuối cùng, hàm lọc lấy các nghiệm thực và loại bỏ nghiệm kép.

  - `sp.solve(expr, symbols)`: Hàm giải đại số tích hợp sẵn của thư viện `symyy`, đóng vai trò là phương án xử lý dự phòng để tìm tập nghiệm chính xác cho các ma trận có đa thức đặc trưng bậc $n > 3$.

- **Ý tưởng thực hiện:**

  1. Từ ma trận đầu vào $M$ cấp $n \times n$ và biến ký hiệu $\lambda$ (symbols), tạo ma trận đặc trưng bằng phép trừ $(M - \lambda I)$, trong đó $I$ là ma trận đơn vị cùng cấp.
  
  2. Sử dụng hàm `det(M)` để tính định thức của ma trận $(M - \lambda I)$ theo công thức khai triển **Laplace** theo dòng đầu tiên $i = 0$. Kết quả thu được là một đa thức đặc trưng $P(\lambda) = \det(M - \lambda I)$.
  
  3. Truyền đa thức đặc trưng $P(\lambda)$ và biến $\lambda$ vào hàm `solvePolynomialEquation(expr, symbols)`. Hàm này sẽ tự động phân loại bậc của đa thức và áp dụng thuật toán tương ứng để giải nghiệm.
  
  4. Tập nghiệm thực tìm được từ đa thức đặc trưng chính là danh sách các trị riêng ($\lambda_1, \lambda_2, \dots$) của ma trận $M$.

### 4.2. Tìm vector riêng & Kiểm tra điều kiện chéo hóa
- `GaussJordan_elimination(M)`: Hàm này nhận vào ma trận $M$ và dùng giải thuật **Gauss-Jordan** để đưa ma trận M về dạng bậc thang rút gọn

$$ M = \begin{pmatrix}
0 & \dots & \mathbf{1} & * & 0 & * & 0 & \dots & * \\
0 & \dots & 0 & 0 & \mathbf{1} & * & 0 & \dots & * \\
0 & \dots & 0 & 0 & 0 & 0 & \mathbf{1} & \dots & * \\
\vdots & & \vdots & & \vdots & & \vdots & \ddots & \vdots \\
0 & \dots & 0 & 0 & 0 & 0 & 0 & \dots & 0
\end{pmatrix} $$

- `back_substitution(M)`: Hàm này nhận vào ma trận mở rộng $M$ có dạng $M = (A|b)$ với $A$ là ma trận đã được đưa về dạng bậc thang hoặc bậc thang rút gọn và $b$ là vector cột các hệ số tự do, sau đó sử dụng phép thế ngược để giải tìm nghiệm của hệ phương trình.

- `null(M)`: Hàm nhận vào ma trận $M = A - \lambda I$ đại diện cho ma trận hệ số của hệ phương trình đồng nhất $(A - \lambda I)X = \mathbf{0}$. Hàm đầu tiên gọi `GaussJordan_elimination(M)` để đưa ma trận về dạng bậc thang rút gọn, sau đó ghép thêm cột hệ số tự do bằng $0$ thông qua `connectMatrixWithCoefficientVector(M, vectorZero)` để tạo ma trận mở rộng $M = (A\vert{}\mathbf{0})$. Cuối cùng, hàm gọi back_substitution(M) thực hiện phép thế ngược để trả về nghiệm tổng quát null_space_expr dưới dạng biểu thức SymPy chứa các biến tự do.

- `getBasisVectors(null_space_expr)`: Hàm nhận vào nghiệm tổng quát null_space_expr để trích xuất ra các vector riêng cơ sở (độc lập tuyến tính). Hàm quét qua các biểu thức để xác định tập hợp các biến tự do (free_vars). Với mỗi biến tự do $x_k$, hàm gán giá trị $x_k = 1$ và tất cả các biến tự do còn lại bằng $0$ (thông qua .subs()), từ đó tạo ra một vector riêng tương ứng. Kết quả trả về là tập hợp các vector cơ sở tạo thành không gian riêng của trị riêng $\lambda$.

- **Ý tưởng thực hiện:**

  1. Với mỗi trị riêng $\lambda_i$ thu được từ đa thức đặc trưng, xây dựng ma trận hệ số $A_i = M - \lambda_i I$.

  2. Tìm không gian nghiệm nullity của $A_i$ tương ứng với trị riêng $\lambda_i$.
       
       - Dùng giải thuật Gauss-Jordan để đưa ma trận $A_i = M - \lambda_i I$ về dạng bậc thang rút gọn

       - Ghép ma trận vừa có được từ giải thuật Gauss-Jordan với vector cột 0 và dùng phép thế ngược để giải ra nghiệm của hệ phương trình tuyến tính thuần nhất $A_i = M - \lambda_i I$.

       - Thu được nghiệm tổng quát `null_space_expr` chính là không gian nghiệm của $A_i$

  3. Truyền `null_space_expr` vào hàm `getBasisVectors` để thu được tập các vector riêng độc lập tuyến tính bằng kỹ thuật gán lần lượt từng biến tự do bằng $1$ và các biến khác bằng $0$. Tất cả các vector riêng của từng trị riêng sẽ được tổng hợp lại vào danh sách chung `combinedBasis`.

  4. Kiểm tra điều kiện chéo hóa và xây dựng kết quả: So sánh tổng số vector riêng độc lập tuyến tính trong `combinedBasis` với cấp $n$ của ma trận gốc $M$:
    
       - Nếu tổng số vector riêng $= n$: Ma trận $M$ chéo hóa được. Tiến hành xếp các vector riêng này làm các cột để tạo nên ma trận chuyển cơ sở $P$, đồng thời lập ma trận đường chéo $D$ chứa các trị riêng tương ứng trên đường chéo chính.
       
       - Nếu tổng số vector riêng $< n$: Ma trận $M$ không thể chéo hóa được (do thiếu vector riêng độc lập tuyến tính).

### 4.3. Xây dựng ma trận $P$, $P^{-1}$ và ma trận chéo $D$
- `getPMatrix(combinedBasis)`:  Hàm nhận vào danh sách các vector riêng cơ sở `combinedBasis` và thực hiện phép chuyển vị để xếp các vector riêng thành các cột, từ đó tạo nên ma trận chuyển cơ sở $P$

- `inverseMatrix(A)`: Hàm nhận vào ma trận $A$ (ở đây là ma trận $P$) và thực hiện giải thuật Gauss-Jordan trên ma trận mở rộng $M = (A | I)$ để tìm ma trận nghịch đảo $A^{-1}$

- `getDiagonalMatrix(P, A, inverseP)`: Hàm nhận vào ma trận chuyển cơ sở $P$, ma trận ban đầu $A$, và ma trận nghịch đảo $P^{-1}$. Hàm thực hiện tính toán phép nhân ma trận liên tiếp $D = P^{-1} A P$ thông qua hàm `multiplyMatrixWithMatrix` để thu về ma trận chéo $D$.

- **Ý tưởng thực hiện:**

  1. Gọi hàm `getPMatrix(combinedBasis)` để ghép các vector riêng độc lập tuyến tính thu được ở bước trước thành các cột của ma trận chuyển cơ sở $P$.

  2. Truyền ma trận $P$ vào hàm `inverseMatrix(P)` để tìm $P^{-1}$ bằng phương pháp biến đổi **Gauss-Jordan** trên ma trận mở rộng $(P | I)$.

  3. Nếu `inverseMatrix` trả về None (ma trận $P$ suy biến, không có nghịch đảo), chương trình đưa ra thông báo ma trận không chéo hóa được và dừng quá trình.

  4. Nếu $P^{-1}$ tồn tại, tiến hành gọi `getDiagonalMatrix(P, A, inverseP)` để thực hiện phép nhân $D = P^{-1} A P$, hoàn tất quá trình chéo hóa ma trận.

```python
A = [[1,-1,-1], 
     [1,3,1], 
     [-3,1,-1]]
P, inverseP, D = diagonalizeMatrix(A)

def printMatrixAligned(M):
    m = len(M[0])
    str_M = [[str(M[i][j]) for j in range(m)] for i in range(n)]
    col_width = [max(len(str_M[i][j]) for i in range(n)) for j in range(m)]
    for i in range(n):
        cells = [str_M[i][j].rjust(col_width[j]) for j in range(m)]
        print("[ " + "   ".join(cells) + " ]")

k = 100

def toRationalMatrix(M):
    return [[sp.nsimplify(val, rational=True) for val in row] for row in M]

P_rat = toRationalMatrix(P)
D_rat = toRationalMatrix(D)
invP_rat = toRationalMatrix(inverseP)

n = len(D_rat)
D_k = [[0] * n for _ in range(n)]
for i in range(n):
    D_k[i][i] = sp.UnevaluatedExpr(sp.Pow(D_rat[i][i], k, evaluate=False))

P_sym = sp.Matrix(P_rat)
D_k_sym = sp.Matrix(D_k)
invP_sym = sp.Matrix(invP_rat)


A_100 = P_sym * D_k_sym * invP_sym

A_100_list = [[A_100[i, j] for j in range(A_100.cols)] for i in range(A_100.rows)]

print("\nMa trận A^100 dưới dạng biểu thức:")
printMatrixAligned(A_100_list)
```

  2. Chuẩn hóa $P$, $D$, $P^{-1}$ về dạng số hữu tỉ chính xác bằng `sp.nsimplify(val, rational=True)`.   3. Xây dựng ma trận đường chéo lũy thừa $D^{100}$ dưới dạng biểu thức:

     - Duyệt qua các phần tử trên đường chéo chính $D[i][i]$ (đã chuẩn hóa về hữu tỉ) và gán $D_k[i][i] = \text{sp.UnevaluatedExpr}(\text{sp.Pow}(D[i][i], k, \text{evaluate=False}))$ với $k = 100$. Việc dùng `sp.Pow(..., evaluate=False)` giúp giữ nguyên biểu thức dạng lũy thừa (ví dụ: $3^{100}, (-2)^{100}$); tuy nhiên chỉ riêng `evaluate=False` là chưa đủ, vì các phép cộng/nhân ma trận ở bước sau vẫn có thể khiến SymPy tự động rút gọn các luỹ thừa số cụ thể thành một con số duy nhất. Bọc thêm `sp.UnevaluatedExpr(...)` để đảm bảo biểu thức luỹ thừa được giữ nguyên xuyên suốt các phép toán ma trận tiếp theo.

  4. Chuyển đổi ma trận $P$, $D_k$, và $P^{-1}$ thành các đối tượng sp.Matrix (P_sym, D_k_sym, invP_sym) để hỗ trợ thực hiện phép đại số biểu thức.

  5. Thực hiện phép nhân ma trận đại số giữa các sp.Matrix. Các phần tử của ma trận $A^{100}$ thu được sẽ là các biểu thức tổng/tích chứa các đại lượng lũy thừa nguyên vẹn, không bị làm tròn hay tính toán ra số thập phân.

## 2. So sánh với module giải từ thư viện sympy

```python
A = sp.Matrix([
    [1, -1, -1],
    [1,  3,  1],
    [-3, 1, -1]
])

def sympyMatrixDiagonalization(A):
    P, D = A.diagonalize()
    inverseP = P.inv()
    print("Ma trận P:")
    sp.pprint(P)
    print("\nMa trận nghịch đảo  P^-1:")
    sp.pprint(inverseP)
    print("\nMa trận đường chéo D:")
    sp.pprint(D)
    return P, inverseP, D
print ("========================= THUẬT TOÁN TỪ THƯ VIỆN SYMPY =========================")
P, inverseP, D = sympyMatrixDiagonalization(A)
```
