# Thông tin giải thuật
-  Giải thuật Gauss là giải thuật dùng để giải hệ phương trình tuyến tính với gồm hai bước chính: biến đổi thành dạng bậc thang cho ma trận mở rộng đầu vào và sử dụng phép thế ngược để tìm nghiệm ra của hệ phương trình. Cụ thể, ma trận hoá hệ phương trình tuyến tính hệ phương trình tính thì được ma trận mở rộng có dạng $\bar{A} = [A|B]$ với A là ma trận các hệ số trước biến và B là cột các hệ số tự do. Sau đó, dùng phép khử Gauss để đưa ma trận mở rộng $\bar{A}$ về dạng bậc thang.

-  Dựa vào dạng bậc thang của $\bar{A}$, suy ra hệ phương trình hiện tại và giải ngược từng ẩn từ phương trình cuối đến phương trình đầu. Các ẩn đã tìm được bên dưới sẽ được đem lên phương trình trên để tiếp tục giải tìm ẩn cho phương trình đang xét. Trường hợp hệ phương trình vô nghiệm là khi dạng bậc thang của $\bar{A}$ tồn tại dòng gồm toàn số 0 nhưng phần tử cuối của dòng đó khác 0 dạng $(0 \quad 0 \quad \ldots \quad  0 \quad | \quad b)$ với $b \neq 0$. Trường hợp hệ có vô nghiệm là khi dạng bậc thang của $\bar{A}$ không chứa dòng vô nghiệm, đồng thời tồn tại dòng gồm toàn số 0 dạng $(0 \quad 0 \quad \ldots \quad 0 \quad | \quad 0)$, hoặc biến ở cột không chứa phần tử chốt được coi là ẩn tự do, khiến cho số phương trình độc lập ít hơn số ẩn của hệ.

```python
import sympy as sp

def relativeCompare(num1, num2, eps = 1e-9):
    return abs(num1 - num2) <= eps

def printMatrix(M):
    n = len(M)
    for i in range(n):
        print(M[i])
```

```python
def swap_row(M, i, j):
    M[i],M[j] = M[j],M[i]

def multiply(M, i, alpha):
    n = len(M[i])
    for x in range (n):
        M[i][x] = M[i][x]*alpha

def eliminate_row(M, i, j, k):
    n = len(M[i])
    for x in range (n):
        val = M[j][x] - k*M[i][x]
        M[j][x] = 0.0 if relativeCompare(val,0) else val
```

```python
def Gauss_elimination(M):
    num_rows = len(M)
    num_cols = len(M[0])
    
    pivot_row = 0
    pivot_col = 0
    
    while pivot_row < num_rows and pivot_col < num_cols:
        # 1. Tìm cột trái nhất không chứa toàn số 0
        nonzero_row = pivot_row
        while nonzero_row < num_rows and relativeCompare(M[nonzero_row][pivot_col],0):
            nonzero_row += 1

        if nonzero_row == num_rows:
            pivot_col += 1
            continue

        # 2. Hoán vị dòng chứa phần tử khác 0 lên dòng pivot hiện tại
        if nonzero_row != pivot_row:
            swap_row(M, pivot_row, nonzero_row)
        
        # 3. Nếu giá trị được kiểm bằng 0 thì gọi multiply để chuẩn hoá về 1
        pivot_val = M[pivot_row][pivot_col]
        if not relativeCompare(pivot_val,1):
            multiply(M, pivot_row, 1/pivot_val)
        
        # 4. Chuẩn hoá giá trị ở các dòng bên dưới trong cột chốt về 0 bằng cách gọi eliminate_row
        for x in range (pivot_row + 1, num_rows):
            if not relativeCompare(M[x][pivot_col], 0):
                eliminate_row(M, pivot_row, x, M[x][pivot_col])
        
        pivot_row += 1
        pivot_col += 1

def GaussJordan_elimination(M):
    num_rows = len(M)
    num_cols = len(M[0])
    
    pivot_row = 0
    pivot_col = 0
    
    while pivot_row < num_rows and pivot_col < num_cols:
        # 1. Tìm cột trái nhất không chứa toàn số 0
        nonzero_row = pivot_row
        while nonzero_row < num_rows and relativeCompare(M[nonzero_row][pivot_col],0):
            nonzero_row += 1

        if nonzero_row == num_rows:
            pivot_col += 1
            continue
        # 2. Hoán vị dòng chứa phần tử khác 0 lên dòng pivot hiện tại
        if nonzero_row != pivot_row:
            swap_row(M, pivot_row, nonzero_row)
        
        # 3. Nếu giá trị được kiểm bằng 0 thì gọi multiply để chuẩn hoá về 1
        pivot_val = M[pivot_row][pivot_col]
        if not relativeCompare(pivot_val,1):
            multiply(M, pivot_row, 1/pivot_val)
        
        # 4. Duyệt tất cả các dòng khác dòng chốt pivot_row trong cột chốt pivot_col để chuẩn hoá về 0
        for x in range (0,num_rows):
            if x == pivot_row:
                continue
            if not relativeCompare(M[x][pivot_col], 0):
                eliminate_row(M, pivot_row, x, M[x][pivot_col])
        
        pivot_row += 1
        pivot_col += 1
```

```python
def back_substitution(M):
    num_rows = len(M)
    num_cols = len(M[0])
    num_vars = num_cols - 1

    result = [None]*num_vars
    
    pivot_positions = {}
    i = num_rows - 1
    # 1. Tìm phần tử cơ sở cho các dòng trong ma trận M
    while i >= 0:
        # 1.1 Tìm phần tử cơ sở trong dòng i
        pivot_col = -1
        for j in range(num_vars):
            if not relativeCompare(M[i][j],0):
                pivot_col = j
                break
        
        # 1.2 Nếu không tìm thấy phần tử cơ sở và phần tử cuối cùng của dòng i khác 0, hệ phương trình vô nghiệm
        if pivot_col == -1 and not relativeCompare(M[i][num_cols - 1],0):
            print("Hệ phương trình vô nghiệm")
            return result
        
        # 1.3 Lưu lại nếu tìm thấy phần tử cơ sở của dòng i
        if pivot_col != -1:
            pivot_positions[i] = pivot_col
        i-=1
        
    # 2. Gán ẩn số thực tự do cho biến tương ứng vớiphần tử không nằm trong pivot_cols
    pivot_cols = set(pivot_positions.values())
    for j in range (num_vars):
        if j not in pivot_cols:
            result[j] = sp.symbols(f'x{j+1}')
    
    i = num_rows - 1
    # 3. Duyệt ngược từ cuối lên đầu để giải hệ phương trình
    while i >= 0:
        # 3.1 Nếu dòng i không có phần tử cơ sở, bỏ qua
        if i not in pivot_positions:
            i-=1
            continue
        # 3.2 Giải phương trình dòng i để tìm ẩn số tương ứng với phần tử cơ sở
        j = pivot_positions[i]

        result[j] = M[i][num_cols - 1]

        for k in range(j + 1, num_vars):
            result[j] -= M[i][k] * result[k]
        i-=1
        
    return result
```

## 1. Tiện ích
- `relativeCompare(num1, num2, eps = 1e-9)`: Hàm này dùng để so sánh bằng giữa hai số $num1$, $num2$ bằng cách so sánh hiệu của chúng xem có bé hơn hoặc bằng sai số cực nhỏ $eps$ cho trước hay không. Thay vì dùng so sánh `==` truyền thống, có thể dẫn đến kết quả sai do vấn đề làm tròn số hoặc biểu diễn dấu chấm động, việc so sánh sử dùng hàm này sẽ đảm bảo kết quả đúng khi thao tác với số thực.

- `printMatrix(M)`: Hàm này sẽ in ra ma trận $M$ được truyền vào với theo format dễ nhìn và cân đối hơn.

- `sympy`: Thư viện này hỗ trợ trong việc tính toán và biểu diễn kết quả cho nghiệm tổng quát trong trường hợp hệ phương trình tuyến tính $Ax = B$ có vô số nghiệm.

## 2. Các phép biến đổi sơ cấp trên dòng
- `swap_row(M, i, j)`: Hàm này nhận 3 tham số, ma trận là $M$, chỉ số hàng i là $i$ và chỉ số hàng j là $j$, và thực hiện hoán vị dòng i và dòng j trong ma trận M.

- `multiply(M, i, alpha)`:  Hàm này nhận 3 tham số, ma trận là $M$, chỉ số hàng i là $i$ và một số thực $alpha$, và thực hiện nhân số thực alpha vào từng phần tử trong dòng i của ma trận M.

- `eliminate_row(M, i, j, k)`: Hàm này 4 tham số, ma trận là $M$, chỉ số hàng i là $i$, chỉ số hàng j là $j$ và hệ số $k$, và thực hiện tính toán giá trị mới cho $dòng_j$ theo công thức: $dòng_j = dòng_j - k\times dòng_i$

## 3. Thuật toán Gauss Elimination
### 3.1. Gauss Elimination
- `Gauss_elimination(M)`: Hàm này nhận vào ma trận M, là ma trận mở rộng có dạng $(A | B)$ với A là ma trận các hệ số trong ma trận và B là cột các hệ số tự do, và biến đổi ma trận M thành ma trận bậc thang có dạng:
$$
\left[
\begin{array}{ccccc|c}
1 & a_{12} & a_{13} & \dots & a_{1n} & b_1 \\
0 & 1 & a_{23} & \dots & a_{2n} & b_2 \\
0 & 0 & 1 & \dots & a_{3n} & b_3 \\
\vdots & \vdots & \vdots & \ddots & \vdots & \vdots \\
0 & 0 & 0 & \dots & 1 & b_n
\end{array}
\right]
$$
- Ý tưởng thực hiện:
  - Tìm ra cột bên trái nhất không chứa toàn bộ là số 0. (1)
    - Nếu dòng chốt $pivot\_ row$ của cột chốt $pivot\_ col$ có giá trị 0 thì tìm hàng khác 0 gần nhất trong cột đang xét
    - Nếu tìm không thấy nghĩa là $nonzero\_ row == num\_ rows$ thì tăng $pivot\_ col$ lên để chuyển sang cột kế tiếp.
  - Trường hợp dòng chốt $pivot\_ row$ của cột đang xét $pivot\_ col$ bằng 0 và tìm được dòng khác 0 trong cùng cột thì gọi hàm `swap()` để hoán vị hai dòng đó.
  - Kiểm tra giá trị ở vị trí dòng chốt $pivot\_ row$ và cột chốt $pivot\_ col$ - phần tử cơ sở $M[pivot\_ row][pivot\_ col]$, nếu khác 0 thì gọi hàm `multiply()` và truyền vào ma trận tham số $alpha = 1/M[pivot\_ row][pivot\_ col]$ và chỉ số dòng chốt $pivot\_ row$ để chuẩn hoá giá trị đó về 1.
  - Từ vị trí $pivot\_ row$ trong cột chốt đang xét $pivot\_ col$, kiểm tra các giá trị của các dòng $x$ bên dưới, nếu giá trị khác 0 thì gọi hàm `eliminate_row()` và truyền vào ma trận, tham số $k = M[x][pivot\_ col]$, chỉ số dòng chốt $pivot\_ row$ và chỉ số dòng $x$ đang xét để chuẩn hoá giá trị đó về 0.
### 3.2. Gauss Jordan Elimination
- `GaussJordan_elimination(M)`: Hàm này nhận vào ma trận M, là ma trận mở rộng có dạng $(A | B)$ với A là ma trận các hệ số trước biến và B là cột các hệ số tự do, và biến đổi ma trận M thành ma trận bậc thang có dạng:
$$
\left[
\begin{array}{ccccc|c}
1 & 0 & 0 & \dots & 0 & b_1 \\
0 & 1 & 0 & \dots & 0 & b_2 \\
0 & 0 & 1 & \dots & 0 & b_3 \\
\vdots & \vdots & \vdots & \ddots & \vdots & \vdots \\
0 & 0 & 0 & \dots & 1 & b_n
\end{array}
\right]
$$
- Ý tưởng thực hiện:
  - Tương tự `Gauss_elimnation()`, chỉ khác bước cuối phải duyệt tất cả các dòng khác dòng chốt pivot\_ row thuộc cột chốt $pivot\_ col$ để đảm bảo tất cả các phần tử đó, ngoại trừ phần tử cơ sở M[pivot\_ row][pivot\_ col], được chuẩn hoá về 0.
## 4. Thuật toán Back substitution
- `back_substitution(M)`: Hàm này nhận vào ma trận mở rộng M đã được đưa về dạng bậc thang bằng `Gauss_elimination(M)` hoặc `GaussJordan_elimination(M)` và giải từ từng phương trình từ dưới lên để tìm nghiệm của hệ phương trình tuyến tính.
- Ý tưởng thực hiện:
  - Tìm phần tử cơ sở của từng dòng trong ma trận $M$
    - Nếu tìm thấy phần tử cơ sở của dòng i đang xét nghĩa là $pivot\_ col \ != -1$ thì lưu lại vị trí của phần tử cơ sở ở dòng i, $pivot\_ positions[i] = pivot\_ col$ và chuyển sang dòng tiếp theo.
    - Nếu không tìm thấy nghĩa là $pivot\_ col == -1$ thì không thực hiện gì thêm và chuyển sang dòng tiếp theo.
    - Trong trường hợp không tìm thấy phần tử cơ sở $pivot\_ col == -1$ và đồng thời phần tử ở cuối dòng i đang xét khác 0 thì hệ phương trình vô nghiệm, trả về mảng có dạng [None, None, $\ldots$, None].
  - Kiểm tra xem ở cột nào thì không có phần tử cơ sở, nếu ở một cột j không có phần tử cơ sở thì gán cho biến tương ứng một ẩn số thực tự do $x_{j+1}$ ($\in \R$) để biểu diễn nghiệm tổng quát cho trường hợp hệ phương trình có vô số nghiệm.
  - Tìm nghiệm cho hệ phương trình tuyến tính, bắt đầu từ phương trình cuối đến phương trình đầu:
    - Nếu dòng i đang xét hiện tại không tồn tại phần tử cơ sở thì chuyển sang dòng tiếp theo.
    - Lấy ra vị trí phần tử cơ sở của dòng $i$ và gán vào $j$. Sau đó, gán nghiệm tương ứng ở vị trí $j$ với giá trị của phần tử cuối cùng của dòng $i$, $result[j] = M[i][num\_ cols - 1]$
    - Nếu đang là phương trình cuối thì không cần tính toán gì thêm, ta tìm được ẩn số đầu tiên.
    - Nếu không phải phương trình cuối thì tình từ vị trí phần tử cơ sở $j$, lấy ra hệ số của các vị trí từ $j + 1$ đến $num\_ vars - 1$ của dòng i $M[i][k]$, và lấy result[j] trừ dần cho $M[i][k]\times result[k]$ với $result[k]$ là các ẩn số tương ứng đã tìm được ở phương trình trước đó.
