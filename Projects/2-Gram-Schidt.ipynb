# Thông tin giải thuật
- **Giải thuận Gram - Schmidt** là một phương pháp toán học kỳ quan trọng trong đại số tuyến tính. Nó dùng để trực giao hóa hoặc trực chuẩn hóa một hệ vector độc lập tuyến tính trong một không gian tích trong, thường gặp nhất là không gian Euclid $\R^n$. Từ một tập vector độc lập tuyến tính bất kì, giải thuật sẽ biến đổi tập đó thành một tập mới có các vector thành phần vuông góc với nhau từng đôi một (trực giao) và có độ dài bằng 1 (trực chuẩn).

- Một trong những ứng dụng của giải thuật Gram-Schmidt là **phép phân rã QR**, phân tích một ma trận $A$ bất kì (nếu có thể) thành tích của một ma trận trực giao $Q$ và một ma trận tam giác trên $R$ ($A = Q \times R$). Từ ứng dụng này, ta có thể giải các hệ phương trình tuyến tính kích thước lớn một cách hiệu quả.

```python
def relativeComparison(a,b,eps = 1e-9):
    return abs(a-b) <= eps
def printMatrix(M):
    for row in M:
        print(row)
def innerProduct(u,v):
    res = 0
    n = len(u)
    if n != len(v):
        return None
    for i in range (n):
        res += u[i]*v[i]
    return res
def length(u):
    return innerProduct(u,u)**0.5
def multiplyWithNumber(a,u):
    n = len(u)
    res = []
    for i in range(n):
        res.append(u[i] * a)
    return res
def sumTwoVector(u,v):
    res = []
    n = len(u)
    if n != len(v):
        return None
    for i in range(n):
        res.append(u[i]+v[i])
    return res
def isVectorZero(u):
    for i in u:
        if not relativeComparison(i,0):
            return False
    return True
def getColumnVector(M,i):
    res = []
    for row in M:
        res.append(row[i])
    return res
def transposeMatrix(M):
    res = []
    cols = len(M[0])
    for i in range(cols):
        res.append(getColumnVector(M,i))
    return res
```

```python
def getMatrixQ(SystemOfColumnVectors):
    n = len(SystemOfColumnVectors)
    if n == 0:
        return None
    
    # Lệ hệ trực giao
    OrthogonalSystem = []
    for i in range(n):
        resVector = SystemOfColumnVectors[i]
        for j in range(len(OrthogonalSystem)):
            subtractedVector = multiplyWithNumber(-innerProduct(OrthogonalSystem[j], SystemOfColumnVectors[i])/(length(OrthogonalSystem[j])**2), OrthogonalSystem[j])
            resVector = sumTwoVector(resVector, subtractedVector)
        if isVectorZero(resVector):
            print("Hệ không độc lập tuyến tính nên không phân rã QR được")
            return None
        OrthogonalSystem.append(resVector)
    

    # Lập hệ trực chuẩn
    OrthonormalSystem = []
    for i in range(n):
        resVector = multiplyWithNumber(1/length(OrthogonalSystem[i]), OrthogonalSystem[i])
        OrthonormalSystem.append(resVector)
    
    # Chuyen vi he truc chuan thanh ma tran Q
    return transposeMatrix(OrthonormalSystem)

# Lập ma trận R
def getMatrixR(SystemOfColumnVectors,MatrixQ):
    res = []
    rowsR = len(MatrixQ[0])
    colsR = len(SystemOfColumnVectors)
    transposedQ = transposeMatrix(MatrixQ)
    for i in range(rowsR):
        resRow = []
        for j in range(colsR):
            R_ij = innerProduct(transposedQ[i], SystemOfColumnVectors[j])
            if relativeComparison(R_ij, 0):
                resRow.append(0)
            else:
                resRow.append(R_ij)
        res.append(resRow)
    return res
```

# Ý tưởng thực hiện và mô tả các hàm
## 1. Tiện ích
- `relativeComparison(a,b,eps = 1e-9)`: Hàm này dùng để so sánh bằng giữa hai số $a$, $b$ bằng cách so sánh hiệu của chúng xem có bé hơn hoặc bằng sai số cực nhỏ $eps$ cho trước hay không. Thay vì dùng so sánh `==` truyền thống, có thể dẫn đến kết quả sai do vấn đề làm tròn số hoặc biểu diễn dấu chấm động, việc so sánh sử dùng hàm này sẽ đảm bảo kết quả đúng khi thao tác với số thực.

- `printMatrix(M)`: Hàm này dùng để in ma trận $M$ theo format đẹp và cân đối hơn.

- `isVectorZero(u)`: Hàm này kiểm tra vector truyển vào $u$ có phải là vector 0 (có dạng $0 = (0, 0, \ldots, 0)$) hay không?

- `getColumnVector(M,i)`: Hàm này dùng để lấy ra vector cột $i$ trong ma trận $M$.

- `tranposeMatrix(M)`: Hàm này dùng để tìm ma trận nghịch đảo của ma trận $M$, cột của ma trận hiện tại trở thành dòng trong ma trận chuyển vị.


## 2. Tính toán trên Vector
- `innerProduct(u,v):` Hàm này dùng để tính tích vô hướng giữa 2 vector truyền vào $u$ và $v$, theo công thức $\langle u, v \rangle = \sum_{i=1}^{n} u_i v_i = u_1 v_1 + u_2 v_2 + \dots + u_n v_n$. Trước khi tính, phải kiểm tra kích thước của hai vector $u$ và $v$ có hợp lệ ($len(u) = len(v)$) để thực hiện việc tính toán không.

- `length(u)`: Hàm này dùng để tính chuẩn của vector $u$, theo công thức $\|u\| = \sqrt{u_1^2 + u_2^2 + \dots + u_n^2}$. Do đã có hàm tính tích vô hướng nên ta có thể tính chuẩn của vector $u$ theo công thức $\|u\| = \sqrt{\langle u, u \rangle}$.

- `multiplyWithNumber(a,u)`: Hàm này dùng để nhân vector $u$ với một số a ($\in \R$), lấy từng phần tử trong trong $u$ nhân với $a$ để tạo ra vector kết quả.

- `sumTwoVector(u,v)`: Hàm này dùng để cộng vector $u$ với vector $v$, lấy từng phần tử u_i$ trong $u$ cộng với từng phần tử $v_i$ ở vị trí tương ứng trong $v$. Trước khi tính, phải kiểm tra kích thước của hai vector $u$ và $v$ có hợp lệ ($len(u) = len(v)$) để thực hiện việc tính toán không.

## 3. Giải thuật Gram - Schmidt tìm ma trận Q
- `getMatrixQ(SystemOfColumnVectors)`: Hàm này nhận vào hệ vector cột *SystemOfColumnVectors* (dạng ($u_1, u_2, \ldots, u_n$)) là chuyển vị của ma trận đầu vào và thực hiện từng bước sau đây để tìm ra ma trận Q:
  - Đầu tiên, lập hệ trực giao với công thức tổng quát:
    $$v_i = u_i - \sum_{j=1}^{i-1} \frac{\langle u_i, v_j \rangle}{\|v_j\|^2} v_j$$
    - Trong đó:
      - $v_i$ và $u_i$: Vector trực giao $i$ cần tìm và vector cột $i$
      - $\langle u_i, v_j \rangle$: Tích vô hướng giữa vector cột $i$ và vector trực giao $j$
      - $\|v_j\|^2$ = $\langle v_j,v_j \rangle$: Chuẩn của vector trực giao $j$
  - Trong quá trình tìm các vector trực giao $v_i$, nếu xuất hiện $v_i = 0$ thì sẽ ngừng thuật toán do hệ vector đầu vào không độc lập tuyến tính nên không thể phân rã QR cho ma trận.
  - Sau khi tìm được hệ trực giao $\text{OrthogonalSystem} = (v_1, v_2,\ldots, v_n)$, tiếp tục tìm hệ trực chuẩn theo công thức tổng quát sau:
    $$q_i = \frac{v_i}{\|v_i\|}$$
    - Trong đó:
      - $q_i$: Vector trực chuẩn $i$
      - $v_i$: Vector trực giao $i$
  - Sau khi tính toán để tìm các vector trực chuẩn, ta có được hệ trực chuẩn $\text{OrthonormalSystem} = (q_1, q_2,\ldots, q_n)$. **Ma trận $Q$** chính là ma trận được tạo thành bằng cách ghép các vectơ trực chuẩn ở dạng cột từ $\text{OrthonormalSystem}$ lại theo thứ tự từ trái sang phải.
## 4. Tìm ma trận R
- `getMatrixR(SystemOfColumnVectors,MatrixQ)`: Hàm này nhận vào hệ vector cột *SystemOfColumnVectors* (dạng ($u_1, u_2, \ldots, u_n$)) là chuyển vị của ma trận đầu vào cùng với ma trận Q (dạng $(q_1, q_2,\ldots, q_n)$) tìm được ở `getMatrixQ(SystemOfColumnVectors)` để tiếp tục tìm ma trận R.

- Phần tử ở hàng i cột j của ma trận R - $R_{ij}$ được tính bằng tích vô hướng giữa vector $q_i$ và $u_j$ $\langle q_i, u_j \rangle$ với $i,j = \overline{1,n}$, ma trận R sẽ có dạng:
 $$R = \begin{pmatrix} 
\langle q_1, u_1 \rangle & \langle q_1, u_2 \rangle & \langle q_1, u_3 \rangle & \dots & \langle q_1, u_n \rangle \\ 
0 & \langle q_2, u_2 \rangle & \langle q_2, u_3 \rangle & \dots & \langle q_2, u_n \rangle \\ 
0 & 0 & \langle q_3, u_3 \rangle & \dots & \langle q_3, u_n \rangle \\ 
\vdots & \vdots & \vdots & \ddots & \vdots \\ 
0 & 0 & 0 & \dots & \langle q_n, u_n \rangle 
\end{pmatrix}$$

# Mở rộng
## 1. Ứng dụng phân rã QR để giải hệ phương trình tuyến tính

```python
# Chuyển vị (MatrixQ) = Nghịch đảo (MatrixQ)
# ùng MatrixQ và MatrixR để giải hệ phương trình tuyến tính Ax = b
# Q^-1 * Q * Rx = Q^-1*b (với A = Q * R) --> Rx = Q^-1*b (với R là ma trận tam giác trên nên gọi back_substitution để giải hệ phương trình)
import sympy as sp
import numpy as np

def multiplyMatrixWithVector(M,v):
    n = len(M[0])
    if n != len(v):
        return None
    res = []
    for row in M:
        res.append(innerProduct(row,v))
    return res

def connectMatrixWithCoefficientVector(M,v):
    n = len(M)
    if n != len(v):
        return None
    for i in range(n):
        M[i].append(v[i])
    return M

def back_substitution(M):
    num_rows = len(M)
    num_cols = len(M[0])
    num_vars = num_cols - 1

    result = [None]*num_vars
    
    pivot_positions = {}
    i = num_rows - 1
    # 1. Tìm phần tử cơ sở cho các dòng trong ma trận M
    while i >= 0:
        # 1.1 Tìm phần tử cơ sở trong dòng i
        pivot_col = -1
        for j in range(num_vars):
            if not relativeComparison(M[i][j],0):
                pivot_col = j
                break
        
        # 1.2 Nếu không tìm thấy phần tử cơ sở và phần tử cuối cùng của dòng i khác 0, hệ phương trình vô nghiệm
        if pivot_col == -1 and not relativeComparison(M[i][num_cols - 1],0):
            print("Hệ phương trình vô nghiệm")
            return result
        
        # 1.3 Lưu lại nếu tìm thấy phần tử cơ sở của dòng i
        if pivot_col != -1:
            pivot_positions[i] = pivot_col
        i-=1
        
    # 2. Gán ẩn số thực tự do cho biến tương ứng với phần tử không nằm trong pivot_cols
    pivot_cols = set(pivot_positions.values())
    for j in range (num_vars):
        if j not in pivot_cols:
            result[j] = sp.symbols(f'x{j+1}')
    
    i = num_rows - 1
    # 3. Duyệt ngược từ cuối lên đầu để giải hệ phương trình
    while i >= 0:
        # 3.1 Nếu dòng i không có phần tử cơ sở, bỏ qua
        if i not in pivot_positions:
            i-=1
            continue
        # 3.2 Giải phương trình dòng i để tìm ẩn số tương ứng với phần tử cơ sở
        j = pivot_positions[i]

        result[j] = M[i][num_cols - 1]

        for k in range(j + 1, num_vars):
            result[j] -= M[i][k] * result[k]
        result[j] /= M[i][j]
        i-=1
    result = [round(x, 9) if isinstance(x, (int, float)) else x for x in result]
    return result

```

```python
A1 = [[1,2,-1],
     [2,2,1],
     [3,5,-2]]
b1 = [-1,1,-1]
systemOfColumnVectors = transposeMatrix(A1)
MatrixQ = getMatrixQ(systemOfColumnVectors)
if MatrixQ is not None:
    MatrixR = getMatrixR(systemOfColumnVectors, MatrixQ)
    InverseMatrixQ = transposeMatrix(MatrixQ)
    coefficientVector = multiplyMatrixWithVector(InverseMatrixQ, b1)
    extendedMatrix = connectMatrixWithCoefficientVector(MatrixR, coefficientVector)
    result = back_substitution(extendedMatrix)
    print(result)
```

### A. Các module liên quan
- `sympy`: Thư viện hỗ trợ biểu diễn nghiệm tổng quát trong trường hợp hệ phương trình tuyến tính $Ax = b$ có vô số nghiệm

- `multiplyMatrixWithVector(M,v)`: Hàm này dùng để nhân ma trận đầu vào $M$ với vector đầu vào $v$ và tạo ra một vector kết quả. Trước khi thực hiện phép tính, phải kiểm tra xem số cột của ma trận $M$ ($len(M[0])$) có bằng số dòng của vector dạng cột $v$ ($len(v)$) không.

- `connectMatrixWithCoefficientVector(M,v)`: Hàm này dùng để ghép ma trận hệ số trước ẩn $M$ và vector cột hệ số tự do $v$ để tạo thành ma trận mở rộng.

- `back_substitution(M)`: Hàm này nhận vào ma trận mở trận mở rộng $M$ và dùng phép thế ngược để giải tìm nghiệm của hệ phương trình tuyến tính (đã được cài đặt ở đồ án Gauss - Elimination).
### B. Ý tưởng thực hiện
- Với yêu cầu giải hệ phương trình tuyến tính $Ax = b$, các bước thực hiện như sau:
  1. **Phân rã QR ma trận A** bằng **thuật toán Gram - Schmidt** và thu được ma trận $Q$ và ma trận $R$.
  2. Với $A = QR$, thay vào vế trái của hệ phương trình tuyến tính ban đầu và có được $QRx = b$ ($\Delta$).
  3. Với tính chất của ma trận trực chuẩn Q là $Q^{-1} = Q^T$, **thực hiện chuyển vị Q** để thu được ma trận Q nghịch đảo $Q^{-1}$.
  4. Nhân trái 2 vế của ($\Delta$) với $Q^{-1}$, thu được $Q^{-1}QRx = Q^{-1}b \ \Leftrightarrow \ Rx = Q^{-1}b$.
  5. Nhân ma trận $Q^{-1}$ với vector cột $b$ và thu được vector cột hệ số tự do mới `coefficientVector`.
  6. Ghép ma trận R và vector cột `coefficientVector` để tạo thành ma trận mở rộng `extendedMatrix`.
  7. Dùng phép thế ngược để giải ma trận mở rộng `extendedMatrix` (ma trận R đã là ma trận tam giác trên) và tìm ra nghiệm của hệ phương trình tuyến tính $Ax = b$.
## 2. So sánh với các module phân rã QR từ thư viện

```python

M = sp.Matrix([[1,1,2],[2,-1,1],[-2,4,1]])

def sympyQRDecomposition(A):
    Q, R = A.QRdecomposition()
    print("Kết quả phân rã QR bằng module từ SymPy:")
    print("Ma trận Q (Đã được trực chuẩn hóa):")
    sp.pprint(Q)

    print("\nMa trận R (Tam giác trên):")
    sp.pprint(R)

def numpyQRDecomposition(A):
    Q, R = np.linalg.qr(A)
    print("Kết quả phân rã QR bằng module từ NumPy:")
    print("Ma trận Q (Đã được trực chuẩn hóa):")
    print(Q)

    print("\nMa trận R (Tam giác trên):")
    print(R)

def selfBuildQRDecompostion(A):
    systemOfColumnVectors = transposeMatrix(A.tolist())
    MatrixQ = getMatrixQ(systemOfColumnVectors)
    if MatrixQ is not None:
        MatrixR = getMatrixR(systemOfColumnVectors, MatrixQ)
        print("Kết quả phân rã QR bằng thuật toán tự xây dựng:")
        print("Ma trận Q (Đã được trực chuẩn hóa):")
        printMatrix(MatrixQ)

        print("\nMa trận R (Tam giác trên):")
        printMatrix(MatrixR)
```

- `QRdecomposition(A)` đến từ `sympy`: Kết quả giống với thuật tự xây dựng do cùng sử dụng giải thuật **Gram - Schmidt** để phân rã QR. Ngoài ra, module từ sympy biểu diễn kết quả ở dạng phân số, đảm bảo độ chính xác đối với kết quả so với cách biểu diễn kết quả dạng thập phân trong thuật toán tự xây dựng.

- `linalg.qr(A)` đến từ `numpy`: Kết quả có chút khác biệt so với thuật toán tự xây dựng do module từ numpy sử dụng giải thuật **Phản xạ Householder**, triệt tiêu các phần tử dưới đường chéo bằng cách phản xạ vector, và hướng phản xạ (dấu cộng hay trừ) được quyết định để tối ưu hóa độ ổn định số học trên máy tính, để phân rã ma trận nhưng kết quả cuối cùng vẫn đảm bảo $A = QR$. Ngoài ra, việc biểu diễn của kết quả của module từ numpy cũng ở dạng thập phân giống với thuật toán tự xây dựng.
